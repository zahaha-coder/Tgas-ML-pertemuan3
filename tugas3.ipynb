```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Muat dataset (pastikan file CSV ada di satu folder atau sesuaikan path-nya)
df = pd.read_csv("spotify_artist_streaming_2020_2025.csv")

# Cek 5 baris pertama data
print("--- 5 Baris Pertama Data ---")
display(df.head()) 

```

```python
# Melihat informasi tipe data dan jumlah baris/kolom
print("--- Informasi Struktur Data ---")
df.info()

# Melihat statistik ringkas (min, max, rata-rata, dll) untuk kolom numerik
print("\n--- Statistik Ringkas ---")
display(df.describe())

# Mengecek apakah ada data yang kosong (missing values)
print("\n--- Cek Data Kosong ---")
print(df.isnull().sum())
```

```python
# Mengisi data kosong pada kolom numerik dengan nilai tengah (median)
for col in df.select_dtypes(include=['number']).columns:
    df[col].fillna(df[col].median(), inplace=True)

# Mengisi data kosong pada kolom objek/kategori dengan modus (nilai yang sering muncul)
for col in df.select_dtypes(include=['object']).columns:
    df[col].fillna(df[col].mode()[0], inplace=True)

# Menghapus baris duplikat jika ada
df.drop_duplicates(inplace=True)

print("Data kosong dan duplikat berhasil dibersihkan!")
print("Sisa total baris data:", len(df))
```

```python
# Mengubah kolom kategori (contoh: 'genre') menjadi angka menggunakan One-Hot Encoding
if 'genre' in df.columns:
    df = pd.get_dummies(df, columns=["genre"], drop_first=True)
    print("Encoding kolom 'genre' selesai!")
    display(df.head())
```

```python
# Menyamakan skala kolom numerik pilihan ke rentang 0 sampai 1 menggunakan MinMaxScaler
scaler = MinMaxScaler()

# Kolom numerik yang relevan di dataset Spotify ini
num_cols = ['danceability', 'energy', 'tempo', 'duration_ms', 'popularity']
existing_num_cols = [col for col in num_cols if col in df.columns]

if existing_num_cols:
    df[existing_num_cols] = scaler.fit_transform(df[existing_num_cols])
    print("Normalisasi numerik ke skala 0-1 berhasil!")
    display(df[existing_num_cols].head())
```

```python
# Membuat fitur baru yang lebih bermakna sesuai materi kuliah (Contoh: Rasio stream terhadap durasi lagu)
if 'stream_count' in df.columns and 'duration_minutes' in df.columns:
    df['streams_per_minute'] = df['stream_count'] / (df['duration_minutes'] + 1)
    print("Fitur baru 'streams_per_minute' berhasil dibuat!")
    display(df[['stream_count', 'duration_minutes', 'streams_per_minute']].head())
```

```python
# Memisahkan data masukan (X) dan target yang ingin diprediksi (y, misal: 'stream_count')
if 'stream_count' in df.columns:
    X = df.drop(columns=['stream_id', 'track_id', 'track_name', 'artist_name', 'album_name', 'release_date', 'stream_count'], errors='ignore')
    y = df['stream_count']
    
    print("Variabel X (Fitur) dan y (Label) berhasil dipisahkan!")
    print("Jumlah fitur X:", X.shape[1])
    print("Jumlah baris y:", len(y))
```
